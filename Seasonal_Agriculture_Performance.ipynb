## 10. Data Cleaning and Preparation

Based on the data quality assessment, the dataset contains a small
number of missing observations in selected numerical variables.

The missing values represent a small proportion of the overall
dataset. Therefore, they will be handled carefully without removing
large portions of the available data.

The following cleaning decisions will be applied:

- Missing rainfall values will be replaced using the median rainfall
  within the corresponding season.
- Missing soil moisture values will be replaced using the median soil
  moisture within the corresponding season.
- Missing yield values will be replaced using the median yield for the
  corresponding crop and season.
- No duplicate records will be removed because no completely duplicate
  records were identified.
- Farm IDs will be retained because all Farm IDs are unique.
- Extreme values will not be removed at this stage. They will be
  investigated separately during the Outlier Analysis section.

### 10.1 Create Clean Dataset

A separate copy of the original dataset is created so that the
original data remains unchanged and the cleaning process can be
tracked.

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Clean dataset shape:", df_clean.shape)
```

### 10.2 Handling Missing Rainfall Values

Rainfall is an important environmental variable in agricultural
analysis. Since rainfall varies considerably across seasons, the
season-wise median is used to replace missing rainfall observations.

Using the median also reduces the influence of extreme rainfall
values.

```python
df_clean["Rainfall_mm"] = df_clean.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median())
)
```

### 10.3 Handling Missing Soil Moisture Values

Soil moisture is another important environmental factor affecting
agricultural performance.

Missing soil moisture values are replaced using the median soil
moisture value within the corresponding season.

```python
df_clean["Soil_Moisture_pct"] = df_clean.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median())
)
```

### 10.4 Handling Missing Yield Values

Yield is one of the primary performance indicators in this project.

Instead of using one overall yield value for all missing observations,
the missing values are replaced using the median yield for the same
crop and season.

This preserves differences in typical crop performance across
different seasonal conditions.

```python
df_clean["Yield_Tonnes_Ha"] = df_clean.groupby(
    ["Crop", "Season"]
)["Yield_Tonnes_Ha"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
remaining_missing = df_clean.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
]

remaining_missing
```

## 12.7 Outlier Analysis

Outliers are observations that lie unusually far from the central distribution of a variable.

The Interquartile Range (IQR) method is used to identify potential outliers. However, an identified outlier is not automatically considered an error. Agricultural variables can naturally contain extreme observations because of differences in weather, farm size, production conditions and other factors.

Therefore, potential outliers are investigated before making any decision to retain or remove them.

```python
# IQR-based outlier summary

outlier_summary = []

for col in important_numeric:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Variable": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": (count / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.round(2)
```

### Outlier Interpretation

The IQR analysis identifies potential outliers across several numerical variables. These observations should not be treated as errors solely because they fall outside the IQR boundaries.

For agricultural data, extreme values may occur because of differences in farm size, rainfall, environmental conditions, crop characteristics, resource usage and production levels.

Therefore, the identified observations are retained unless there is sufficient evidence that they are incorrect or represent data-entry errors.

Correlation analysis is used to examine the strength and direction of

The correlation coefficient ranges from -1 to +1. Positive values

### 15.5 Strongest Correlations

The strongest correlations among the selected numerical variables are
identified to focus the interpretation on the most meaningful
relationships in the dataset.

```python
# 15.5 Strongest correlations

corr_pairs = (
    correlation_matrix
    .where(~np.eye(correlation_matrix.shape[0], dtype=bool))
    .stack()
    .reset_index()
)

corr_pairs.columns = ["Variable_1", "Variable_2", "Correlation"]

corr_pairs["Abs_Correlation"] = corr_pairs["Correlation"].abs()

strongest_correlations = (
    corr_pairs
    .sort_values("Abs_Correlation", ascending=False)
    .drop_duplicates(subset=["Abs_Correlation"])
    .head(10)
)

strongest_correlations.round(2)
```

### 17.1 Which Season Has Better Water-Use Efficiency?

Water usage alone does not indicate efficient agricultural
performance. Therefore, yield is compared with water usage to
evaluate water-use efficiency across seasons.

```python
# 17.1 Seasonal Water-Use Efficiency

season_efficiency = (
    df.groupby("Season")
    .agg(
        Average_Yield=("Yield_Tonnes_Ha", "mean"),
        Average_Water_Used=("Water_Used_m3", "mean")
    )
)

season_efficiency["Yield_per_1000m3"] = (
    season_efficiency["Average_Yield"] /
    season_efficiency["Average_Water_Used"] * 1000
)

season_efficiency.round(2)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=season_efficiency.index,
    y=season_efficiency["Yield_per_1000m3"]
)

plt.title("Seasonal Water-Use Efficiency")
plt.xlabel("Season")
plt.ylabel("Yield per 1000 m³ of Water")
plt.tight_layout()
plt.show()
```

### Observation

The calculated efficiency compares agricultural yield with the amount
of water used.

A higher value indicates that more yield was obtained per 1000 m³ of
water used. This provides a more meaningful comparison of water
utilization than examining total water usage alone.
